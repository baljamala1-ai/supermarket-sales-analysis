## OOP and Exception Handling
bold text             


In this part, each sale transaction is represented as an object.  
A base class called SaleTransaction is created, and two child classes are created: MemberSale and NormalSale.  
A custom exception is used to reject invalid transaction data.

```python
class InvalidTransactionError(Exception):
    pass
```

```python
class SaleTransaction:
    def __init__(self, invoice_id, customer_type, product_line, unit_price, quantity, sales, rating):

        if unit_price <= 0:
            raise InvalidTransactionError("Unit price must be greater than zero.")

        if quantity <= 0:
            raise InvalidTransactionError("Quantity must be greater than zero.")

        if pd.isna(product_line) or product_line == "":
            raise InvalidTransactionError("Product line cannot be missing.")

        if rating < 0 or rating > 10:
            raise InvalidTransactionError("Rating must be between 0 and 10.")

        self.invoice_id = invoice_id
        self.customer_type = customer_type
        self.product_line = product_line

        self.__unit_price = unit_price
        self.__quantity = quantity

        self.sales = sales
        self.rating = rating

    def calculate_subtotal(self):
        return self.__unit_price * self.__quantity

    def get_unit_price(self):
        return self.__unit_price

    def get_quantity(self):
        return self.__quantity

    def sale_summary(self):
        return f"Invoice {self.invoice_id}: {self.customer_type} customer bought {self.product_line} with sales amount {self.sales:.2f}"
```

```python
class MemberSale(SaleTransaction):
    def calculate_final_sales(self):
        return self.sales * 0.95

    def sale_summary(self):
        return f"Member invoice {self.invoice_id}: final sales after 5% discount is {self.calculate_final_sales():.2f}"
```

```python
class NormalSale(SaleTransaction):
    def calculate_final_sales(self):
        return self.sales

    def sale_summary(self):
        return f"Normal invoice {self.invoice_id}: sales amount is {self.calculate_final_sales():.2f}"
```

```python
transactions = []

for index, row in df.head(15).iterrows():
    try:
        if row["Customer type"] == "Member":
            transaction = MemberSale(
                row["Invoice ID"],
                row["Customer type"],
                row["Product line"],
                row["Unit price"],
                row["Quantity"],
                row["Sales"],
                row["Rating"]
            )
        else:
            transaction = NormalSale(
                row["Invoice ID"],
                row["Customer type"],
                row["Product line"],
                row["Unit price"],
                row["Quantity"],
                row["Sales"],
                row["Rating"]
            )

        transactions.append(transaction)

    except InvalidTransactionError as error:
        print(f"Invalid transaction skipped: {error}")
```

```python
for transaction in transactions:
    print(transaction.sale_summary())
```

# Milestone 4: Pandas Data Cleaning, Analysis & Excel Export

In this final milestone, we clean the dataset by standardizing column names, handling duplicates, and parsing dates. We then answer key analytical business questions, generate a Pivot Table, and export the structured report to an Excel file.

## Part 1: Data Cleaning & Preprocessing

In this section, we clean and standardize the dataset by stripping column whitespace, removing duplicate records, handling missing values, and formatting the `Date` column into a proper datetime structure.

```python
# 1. Clean column names (strip whitespace and standardise)
df.columns = df.columns.str.strip()
```

```python
# 2. Handle missing values & remove duplicates
df.drop_duplicates(inplace=True)
df.fillna(method='ffill', inplace=True)
```

```python
# 3. Convert Date column to datetime format
df['Date'] = pd.to_datetime(df['Date'])
```

```python
print("Data Cleaning Complete. Missing values:", df.isnull().sum().sum())
```

## Part 2: Analytical Questions, Pivot Table & Excel Export

In this section, we process business queries to derive key analytical insights, construct a Pivot Table for cross-branch performance comparison, and export the structured results into an Excel report.

```python
# Question 1: Which branch achieved the highest total sales?
top_branch = df.groupby('Branch')['Sales'].sum().idxmax()
top_branch_sales = df.groupby('Branch')['Sales'].sum().max()
```

```python
# Question 2: Which product line has the highest average rating?
top_rated_product = df.groupby('Product line')['Rating'].mean().idxmax()
top_rating_score = df.groupby('Product line')['Rating'].mean().max()
```

```python
# Question 3: What is the most used payment method?
most_frequent_payment = df['Payment'].value_counts().idxmax()
```

```python
print("=== Analytical Results ===")
print(f"1. Top Performing Branch        : {top_branch} (${top_branch_sales:,.2f})")
print(f"2. Top Rated Product Line       : {top_rated_product} ({top_rating_score:.2f}/10)")
print(f"3. Most Used Payment Method     : {most_frequent_payment}")
```

```python
# Pivot Table Analysis (Sales by Branch and Product Line)
pivot_sales = pd.pivot_table(df, values='Sales', index='Product line', columns='Branch', aggfunc='sum')
print("\n=== Branch vs Product Line Pivot Table ===")
print(pivot_sales)
```

```python
import pandas as pd

# 1. Aggregate analytical question results into a DataFrame
analysis_summary_df = pd.DataFrame([
    {"Metric": "Top Performing Branch", "Value": f"{top_branch} (${top_branch_sales:,.2f})"},
    {"Metric": "Top Rated Product Line", "Value": f"{top_rated_product} ({top_rating_score:.2f}/10)"},
    {"Metric": "Most Used Payment Method", "Value": most_frequent_payment}
])

# 2. Export all three sheets to the Excel file
file_name = "Supermarket_Sales_Report.xlsx"

with pd.ExcelWriter(file_name, engine='openpyxl') as writer:
    df.to_excel(writer, sheet_name='Cleaned Data', index=False)
    pivot_sales.to_excel(writer, sheet_name='Sales Summary')
    analysis_summary_df.to_excel(writer, sheet_name='Analysis Summary', index=False)

print(f"Successfully exported report to '{file_name}'!")
```
